# All Metrics

```python
def get_metrics(config: dict):
    metrics_path = "/home/victor-hugo/HAVANA-2.0/data/processed/metrics/"

    # Baseline metrics
    recall_baseline = pd.read_csv(
        f"/home/victor-hugo/HAVANA-2.0/data/processed/metrics/baseline/{config['state']}/recall_baseline.csv"
    )

    precision_baseline = pd.read_csv(
        f"/home/victor-hugo/HAVANA-2.0/data/processed/metrics/baseline/{config['state']}/precision_baseline.csv"
    )

    fscore_baseline = pd.read_csv(
        f"/home/victor-hugo/HAVANA-2.0/data/processed/metrics/baseline/{config['state']}/fscore_baseline.csv"
    )

    # region embeddings metrics
    recall_metrics = pd.read_csv(
        metrics_path + 
        f"{config['embedder']}/{config['state']}/recall_"
        + f"{config['embeddings_dimension']}_dimension_{config['h3_resolution']}_resolution.csv"
    )

    precision_metrics = pd.read_csv(
        metrics_path + 
        f"{config['embedder']}/{config['state']}/precision_"
        + f"{config['embeddings_dimension']}_dimension_{config['h3_resolution']}_resolution.csv"
    )

    fscore_metrics = pd.read_csv(
        metrics_path + 
        f"{config['embedder']}/{config['state']}/fscore_"
        + f"{config['embeddings_dimension']}_dimension_{config['h3_resolution']}_resolution.csv"
    )

    print("Diferença")
    print(f"Recall: {(recall_metrics.mean() - recall_baseline.mean()).sum()}")
    print(f"Precision: {(precision_metrics.mean() - precision_baseline.mean()).sum()}")
    print(f"F1-Score: {(fscore_metrics.mean() - fscore_baseline.mean()).sum()}")
    
    print("Proporção")
    print(f"Recall: {(recall_metrics.mean() / recall_baseline.mean()).mean()}")
    print(f"Precision: {(precision_metrics.mean() / precision_baseline.mean()).mean()}")
    print(f"F1-Score: {(fscore_metrics.mean() / fscore_baseline.mean()).mean()}")
```

```python
config={
    "state": "Texas",
    "embedder": "poi-encoder",
    "embeddings_dimension": 32,
    "h3_resolution": 8
}
# Baseline metrics
recall_baseline = pd.read_csv(
    f"/home/victor-hugo/HAVANA-2.0/data/processed/metrics/baseline/{config['state']}/recall_baseline.csv"
)

precision_baseline = pd.read_csv(
    f"/home/victor-hugo/HAVANA-2.0/data/processed/metrics/baseline/{config['state']}/precision_baseline.csv"
)

fscore_baseline = pd.read_csv(
    f"/home/victor-hugo/HAVANA-2.0/data/processed/metrics/baseline/{config['state']}/fscore_baseline.csv"
)
metrics_path = "/home/victor-hugo/HAVANA-2.0/data/processed/metrics/"
# region embeddings metrics
recall_metrics = pd.read_csv(
    metrics_path + 
    f"{config['embedder']}/{config['state']}/recall_"
    + f"{config['embeddings_dimension']}_dimension_{config['h3_resolution']}_resolution.csv"
)

precision_metrics = pd.read_csv(
    metrics_path + 
    f"{config['embedder']}/{config['state']}/precision_"
    + f"{config['embeddings_dimension']}_dimension_{config['h3_resolution']}_resolution.csv"
)

fscore_metrics = pd.read_csv(
    metrics_path + 
    f"{config['embedder']}/{config['state']}/fscore_"
    + f"{config['embeddings_dimension']}_dimension_{config['h3_resolution']}_resolution.csv"
)

print(f"Recall: {(recall_metrics.mean() - recall_baseline.mean()).sum()}")
print(f"Precision: {(precision_metrics.mean() - precision_baseline.mean()).sum()}")
print(f"F1-Score: {(fscore_metrics.mean() - fscore_baseline.mean()).sum()}")
```
